````python
# React Interview Roadmap — Full Guide

---

## Phase 1: React Fundamentals

### 1. What is React?
A JavaScript library (not a full framework) for building user interfaces, maintained by Meta. It lets you build UIs as a tree of
reusable, declarative **components** — you describe *what* the UI should look like for a given state, and
React figures out *how* to update the DOM to match.

### 2. Why React?
- **Declarative**: describe the end state, not step-by-step DOM manipulation.
- **Component-based**: UI split into independent, reusable pieces.
- **Virtual DOM**: fast, batched updates instead of expensive direct DOM writes.
- **Huge ecosystem**: routing, state management, testing, meta-frameworks (Next.js) all built around it.
- **Learn once, write anywhere**: same mental model powers React Native for mobile.

### 3. Features of React
- JSX syntax
- Component-based architecture
- Virtual DOM & efficient reconciliation
- One-way (unidirectional) data binding
- Hooks for state/lifecycle in function components
- Strong ecosystem (Router, Redux, React Query, Next.js)
- Server-side rendering & static generation support (via frameworks)

### 4. Virtual DOM
An in-memory, lightweight JS representation of the real DOM tree. When state changes, React builds a new Virtual
DOM tree, **diffs** it against the previous one, and applies only the minimal set of changes to the actual DOM.
This avoids costly full re-renders of the real DOM, which is much slower to touch than plain JS objects.

### 5. Real DOM vs Virtual DOM

| Real DOM | Virtual DOM |
|---|---|
| Updating it directly is slow (triggers layout/reflow/repaint) | Just a JS object — cheap to create and compare |
| Updates the whole tree/branch on change | React calculates the minimal diff before touching real DOM |
| Manipulated directly by vanilla JS/jQuery | Managed internally by React; you never touch it directly |

### 6. JSX
JSX (JavaScript XML) lets you write HTML-like syntax inside JS. It's not valid JS by itself — Babel compiles it to `React.createElement()` calls.
```jsx
const element = <h1 className="greeting">Hello, {name}!</h1>;
// compiles to:
const element = React.createElement('h1', { className: 'greeting' }, `Hello, ${name}!`);
```
Rules: must return a single root element (or Fragment `<>...</>`), use `className` not `class`, and JS expressions go inside `{}`.

### 7. Components
The building blocks of a React UI — independent, reusable pieces that each return JSX. Two kinds: **function components**
(modern standard) and **class components** (legacy). Components can be composed: a parent renders children, which render
their own children, forming a tree.

### 8. Functional vs Class Components
```jsx
// Function component (modern)
function Welcome({ name }) {
  const [count, setCount] = useState(0);
  return <h1>Hello, {name} ({count})</h1>;
}

// Class component (legacy)
class Welcome extends React.Component {
  state = { count: 0 };
  render() {
    return <h1>Hello, {this.props.name} ({this.state.count})</h1>;
  }
}
```
Function components use Hooks (`useState`, `useEffect`, etc.) instead of `this.state` and lifecycle methods.
They're shorter, avoid `this`-binding issues, and are the recommended approach in all current React code.

### 9. Props
Short for "properties" — read-only data passed from a parent component to a child. A component must never mutate its own
props; that's what breaks the one-way data flow.
```jsx
function Greeting({ name, age }) {
  return <p>{name} is {age} years old</p>;
}
<Greeting name="Vijay" age={26} />
```

### 10. State
Local, mutable data owned by a component, managed via `useState`/`useReducer` (function components) or `this.state` (class components). Changing state triggers a re-render. Unlike props, a component fully controls its own state.

### 11. One-way Data Binding
Data flows in a single direction: parent → child, via props. A child cannot directly change a prop it receives; to affect parent state, the parent passes a **callback function** down as a prop, and the child calls it. This "props down, events up" pattern makes data flow predictable and easy to trace, unlike two-way binding frameworks (e.g. older Angular) where child changes could silently mutate parent state.

---

## Phase 2: Component Lifecycle & Hooks

### 12. Rendering Process
1. **Trigger**: state/prop change, parent re-render, or context change.
2. **Render**: React calls the component function, builds a new Virtual DOM tree.
3. **Reconciliation**: React diffs the new tree against the previous one.
4. **Commit**: React applies the minimal set of real DOM mutations.
5. **Browser paint**: the updated UI becomes visible.

Class components expose this as named lifecycle methods (`componentDidMount` → mount, `componentDidUpdate` → update, `componentWillUnmount` → unmount); function components achieve the same via `useEffect`.

### 13. useState
```jsx
const [count, setCount] = useState(0);
setCount(count + 1);          // direct update
setCount(prev => prev + 1);   // functional update — safe when new state depends on old
```
Calling the setter schedules a re-render; multiple `setState` calls in the same event handler are batched into one re-render.

### 14. useEffect
Runs side effects (data fetching, subscriptions, timers, manual DOM work) after render.
```jsx
useEffect(() => {
  const id = setInterval(tick, 1000);
  return () => clearInterval(id); // cleanup: runs before next effect and on unmount
}, [dep]); // re-runs only when `dep` changes; [] = run once on mount
```

### 15. useRef
Returns a mutable `.current` object that persists across renders **without** causing a re-render when changed. Two main uses:
```jsx
// 1. DOM access
const inputRef = useRef(null);
useEffect(() => inputRef.current.focus(), []);
<input ref={inputRef} />

// 2. Mutable instance value (e.g. storing a previous value or a timer id)
const renderCount = useRef(0);
renderCount.current += 1;
```

### 16. useMemo
Memoizes a **computed value**, recalculating only when its dependencies change — used to avoid expensive recalculations on every render.
```jsx
const sortedList = useMemo(() => expensiveSort(items), [items]);
```

### 17. useCallback
Memoizes a **function reference**, so it doesn't get recreated (and doesn't break `React.memo` on children) on every render.
```jsx
const handleClick = useCallback(() => doSomething(id), [id]);
```
`useCallback(fn, deps)` is functionally equivalent to `useMemo(() => fn, deps)`.

### 18. Custom Hooks
A plain function, name prefixed with `use`, that calls other Hooks internally — a way to extract and reuse stateful logic across components.
```jsx
function useFetch(url) {
  const [data, setData] = useState(null);
  useEffect(() => {
    fetch(url).then(r => r.json()).then(setData);
  }, [url]);
  return data;
}
```

### 19. Rules of Hooks
1. **Only call Hooks at the top level** — never inside loops, conditions, or nested functions.
2. **Only call Hooks from React function components or other custom Hooks** — not plain JS functions.
Reason: React associates Hook state with a component instance purely by **call order**; breaking that order across renders corrupts state.

---

## Phase 3: Event Handling & Forms

### 20. Events
React wraps native DOM events in a cross-browser `SyntheticEvent` for consistent behavior, and uses camelCase handler names.
```jsx
<button onClick={(e) => console.log(e.target)}>Click</button>
```

### 21. Controlled Components
Form element value is driven entirely by React state; every change goes through `onChange`, so React state is always the single source of truth.
```jsx
function ControlledInput() {
  const [value, setValue] = useState('');
  return <input value={value} onChange={e => setValue(e.target.value)} />;
}
```

### 22. Uncontrolled Components
Form element manages its own value in the DOM; React reads it on demand via a `ref` instead of tracking it in state on every keystroke.
```jsx
function UncontrolledInput() {
  const inputRef = useRef();
  const handleSubmit = () => alert(inputRef.current.value);
  return <input ref={inputRef} defaultValue="" />;
}
```
Trade-off: controlled gives full control/validation on every keystroke but re-renders more; uncontrolled is simpler and closer to plain HTML but harder to validate live.

### 23. Form Validation
Typically done either manually (checking values in state/on submit) or with a library (Formik, React Hook Form + Yup/Zod schemas).
```jsx
function SignupForm() {
  const [email, setEmail] = useState('');
  const [error, setError] = useState('');

  const handleSubmit = (e) => {
    e.preventDefault();
    if (!/\S+@\S+\.\S+/.test(email)) {
      setError('Invalid email');
      return;
    }
    setError('');
    // submit
  };

  return (
    <form onSubmit={handleSubmit}>
      <input value={email} onChange={e => setEmail(e.target.value)} />
      {error && <p>{error}</p>}
      <button type="submit">Submit</button>
    </form>
  );
}
```

---

## Phase 4: Routing (React Router)

### 24. React Router
The standard client-side routing library for React SPAs — maps URL paths to components without a full page reload.

### 25. BrowserRouter
Wraps the app and enables routing using the HTML5 History API (clean URLs, no `#`).
```jsx
import { BrowserRouter } from 'react-router-dom';
<BrowserRouter><App /></BrowserRouter>
```

### 26. Routes
Container that renders the first `<Route>` matching the current URL (React Router v6+).
```jsx
import { Routes, Route } from 'react-router-dom';
<Routes>
  <Route path="/" element={<Home />} />
  <Route path="/about" element={<About />} />
</Routes>
```

### 27. Route
Defines a single path-to-component mapping, as shown above. Supports nested routes and dynamic segments (`/users/:id`).

### 28. Link
Renders an `<a>` tag that navigates client-side (no full page reload) — always use instead of a plain `<a href>` for internal navigation.
```jsx
import { Link } from 'react-router-dom';
<Link to="/about">About</Link>
```

### 29. NavLink
Like `Link`, but automatically applies an `active` class (or a custom style via a function) when its `to` path matches the current URL — used for nav menus.
```jsx
<NavLink to="/about" className={({ isActive }) => isActive ? 'active' : ''}>About</NavLink>
```

### 30. useNavigate
Hook for programmatic navigation (e.g., after a form submits or a redirect).
```jsx
const navigate = useNavigate();
const handleLogin = () => { /* ... */ navigate('/dashboard'); };
```

### 31. URL Parameters
Dynamic segments in a route path, read via `useParams`.
```jsx
<Route path="/users/:id" element={<UserProfile />} />

function UserProfile() {
  const { id } = useParams();
  return <p>User ID: {id}</p>;
}
```

---

## Phase 5: State Management

### 32. Lifting State Up
When two sibling components need to share/sync state, move that state to their closest common ancestor and pass it down via props (plus a callback for the children to update it).

### 33. Context API
Built-in way to share data across the component tree without prop drilling.
```jsx
const ThemeContext = createContext('light');

function App() {
  return (
    <ThemeContext.Provider value="dark">
      <Toolbar />
    </ThemeContext.Provider>
  );
}
function Toolbar() {
  const theme = useContext(ThemeContext);
  return <div className={theme}>...</div>;
}
```
Caveat: every consumer re-renders when the provided value changes, even if it only needs part of it — fine for low-frequency global data (theme, auth user, locale), less ideal for high-frequency state.

### 34. Redux Basics
A predictable, centralized state container: a single **store** holds app state; components **dispatch actions** (plain objects describing "what happened"); pure **reducer** functions compute the new state from the action.
```jsx
// reducer
function counterReducer(state = { count: 0 }, action) {
  switch (action.type) {
    case 'increment': return { count: state.count + 1 };
    default: return state;
  }
}
// dispatch
store.dispatch({ type: 'increment' });
```
Core principles: single source of truth, state is read-only (only changed via dispatched actions), changes are made with pure functions.

### 35. Redux Toolkit
The modern, official way to write Redux — cuts boilerplate via `configureStore`, `createSlice`, and built-in Immer (lets you "mutate" state in reducers while it's actually handled immutably under the hood).
```jsx
import { createSlice, configureStore } from '@reduxjs/toolkit';

const counterSlice = createSlice({
  name: 'counter',
  initialState: { count: 0 },
  reducers: {
    increment: (state) => { state.count += 1; }, // looks mutable, is safe (Immer)
  },
});

const store = configureStore({ reducer: { counter: counterSlice.reducer } });
export const { increment } = counterSlice.actions;
```

---

## Phase 6: API Calls

### 36. Fetch API
Native browser API for HTTP requests, returns a Promise.
```jsx
fetch('/api/users')
  .then(res => res.json())
  .then(data => console.log(data))
  .catch(err => console.error(err));
```

### 37. Axios
Popular third-party HTTP client — auto JSON-parses responses, has cleaner error handling (rejects on non-2xx status, unlike `fetch`), supports interceptors and request cancellation.
```jsx
import axios from 'axios';
const { data } = await axios.get('/api/users');
```

### 38. Async/Await
Cleaner syntax over `.then()` chains for handling Promises inside `useEffect` (note: the effect callback itself can't be `async` directly, so wrap it):
```jsx
useEffect(() => {
  async function loadData() {
    const res = await fetch('/api/users');
    const data = await res.json();
    setUsers(data);
  }
  loadData();
}, []);
```

### 39. Loading States
Track request status explicitly so the UI can show a spinner, the data, or an error correctly.
```jsx
const [status, setStatus] = useState('idle'); // idle | loading | success | error
```

### 40. Error Handling
Wrap requests in try/catch (or `.catch()`), guard against setting state after unmount, and surface errors to the user.
```jsx
useEffect(() => {
  let cancelled = false;
  setStatus('loading');
  fetch('/api/users')
    .then(res => { if (!res.ok) throw new Error('Request failed'); return res.json(); })
    .then(data => !cancelled && setStatus('success'))
    .catch(() => !cancelled && setStatus('error'));
  return () => { cancelled = true; };
}, []);
```

---

## Phase 7: Performance Optimization

### 41. React.memo
Wraps a function component so it skips re-rendering if its props are shallow-equal to the previous render.
```jsx
const UserRow = React.memo(function UserRow({ user }) {
  return <li>{user.name}</li>;
});
```

### 42. useMemo
(See #16) — memoize expensive computed values so they're only recalculated when dependencies change.

### 43. useCallback
(See #17) — memoize function references, especially useful when passing callbacks to `React.memo`-wrapped children.

### 44. Lazy Loading
Load a component's code only when it's needed, reducing the initial bundle size.
```jsx
const Chart = React.lazy(() => import('./Chart'));
```

### 45. Suspense
Declaratively shows a fallback UI while a lazily-loaded component (or, in newer React, a data-fetching component) is still loading.
```jsx
<Suspense fallback={<Spinner />}>
  <Chart />
</Suspense>
```

### 46. Code Splitting
The broader strategy of breaking your bundle into smaller chunks loaded on demand — usually
 implemented via `React.lazy` per route (so each page's code loads only when the user navigates there) combined with a bundler (Webpack/Vite) that automatically creates the split chunks.
```jsx
const Home = React.lazy(() => import('./pages/Home'));
const About = React.lazy(() => import('./pages/About'));

<Suspense fallback={<Spinner />}>
  <Routes>
    <Route path="/" element={<Home />} />
    <Route path="/about" element={<About />} />
  </Routes>
</Suspense>
```

---

## Quick Self-Check
Before an interview, you should be able to, without notes:
- Explain reconciliation and why keys matter, from memory.
- Write a controlled form with validation from scratch.
- Set up a route with a dynamic param and read it with `useParams`.
- Explain when you'd reach for Context vs Redux Toolkit vs just lifting state up.
- Write a `useEffect` data-fetch with loading/error states and proper cleanup.
- Explain the difference between `useMemo`, `useCallback`, and `React.memo` in one sentence each.
````

````python
# Phase 3 – Modules (Deep Dive)

Modules help Angular organize an application into logical groups of related functionality. While **modern Angular (v17+) encourages standalone components**, **NgModules are still widely used in enterprise applications**, so interviewers frequently ask about them.

---

# What is a Module?

A **Module** is a class decorated with `@NgModule` that groups related components, directives, pipes, and services.

Think of an online shopping application.

```text
Shopping Application

├── Login Module
├── Product Module
├── Cart Module
├── Order Module
├── Payment Module
└── Admin Module
```

Each module is responsible for one feature.

---

# Why Do We Need Modules?

Imagine an application with:

* 100 Components
* 30 Services
* 20 Directives
* 10 Pipes

Without modules:

```text
Everything inside one folder

AppComponent
LoginComponent
CartComponent
ProductComponent
PaymentComponent
...
100+ files together
```

It becomes difficult to maintain.

With modules:

```text
Login Module

├── LoginComponent
├── RegisterComponent
└── ForgotPasswordComponent
```

```text
Product Module

├── ProductListComponent
├── ProductDetailsComponent
└── ProductCardComponent
```

Much cleaner and easier to manage.

---

# AppModule

Every Angular application traditionally starts with an **AppModule**.

Example:

```typescript
import { NgModule } from '@angular/core';
import { BrowserModule } from '@angular/platform-browser';

import { AppComponent } from './app.component';

@NgModule({

  declarations: [
    AppComponent
  ],

  imports: [
    BrowserModule
  ],

  providers: [],

  bootstrap: [
    AppComponent
  ]

})

export class AppModule { }
```

---

# Understanding @NgModule

This is one of the most commonly asked interview questions.

```typescript
@NgModule({

declarations: [],

imports: [],

providers: [],

bootstrap: []

})
```

Let's understand every property.

---

# declarations

Contains the components, directives, and pipes that belong to this module.

Example

```typescript
declarations: [

AppComponent,

LoginComponent,

HomeComponent

]
```

Think of it as:

> "These are my UI building blocks."

---

# imports

Imports functionality from other modules.

Example

```typescript
imports: [

BrowserModule,

FormsModule,

HttpClientModule

]
```

Without importing a module, its features cannot be used.

Example:

Without `FormsModule`

```html
<input [(ngModel)]="name">
```

Angular throws an error because `ngModel` belongs to `FormsModule`.

---

# providers

Registers services for Dependency Injection.

Example

```typescript
providers:[

EmployeeService

]
```

Angular creates and manages service instances for injection.

> In modern Angular, many services use `providedIn: 'root'`, so you often don't need to list them here.

---

# bootstrap

Specifies the first component Angular loads.

Example

```typescript
bootstrap:[

AppComponent

]
```

Angular starts rendering from `AppComponent`.

---

# Types of Modules

## 1. Root Module

The first module of the application.

Example:

```text
AppModule
```

Only one root module exists in a traditional Angular application.

---

## 2. Feature Module

Groups related business features.

Example

```text
AdminModule

EmployeeModule

OrderModule

PaymentModule
```

---

## 3. Shared Module

Contains reusable items.

Example

```text
NavbarComponent

FooterComponent

CustomPipe

CommonDirectives
```

Imported wherever needed.

---

## 4. Core Module

Contains singleton services used throughout the application.

Example

```text
AuthenticationService

LoggingService

UserService
```

Usually imported only once by the root module.

---

# BrowserModule vs CommonModule

This is a favorite interview question.

| BrowserModule                      | CommonModule                                         |
| ---------------------------------- | ---------------------------------------------------- |
| Used in the root module            | Used in feature modules                              |
| Required once                      | Can be imported many times                           |
| Includes browser-specific services | Provides common directives like `*ngIf` and `*ngFor` |

Example

Root Module

```typescript
imports:[
BrowserModule
]
```

Feature Module

```typescript
imports:[
CommonModule
]
```

---

# FormsModule

Used for template-driven forms.

Example

```typescript
imports:[

FormsModule

]
```

Then you can write

```html
<input [(ngModel)]="username">
```

---

# ReactiveFormsModule

Used for reactive forms.

```typescript
imports:[

ReactiveFormsModule

]
```

Preferred for large enterprise applications because it offers better scalability and validation.

---

# HttpClientModule

Required for REST API calls.

```typescript
imports:[

HttpClientModule

]
```

Without it, Angular's `HttpClient` cannot be injected.

---

# RouterModule

Provides routing support.

```typescript
imports:[

RouterModule

]
```

Used for navigation between pages.

---

# Exporting Components

Suppose

```text
Shared Module

↓

Navbar Component
```

Other modules need it.

```typescript
exports:[

NavbarComponent

]
```

Now any module importing `SharedModule` can use `NavbarComponent`.

---

# Lazy Loaded Modules

Instead of loading everything at startup,

Angular loads modules only when required.

Example

Without Lazy Loading

```text
Application Starts

↓

Load Login

↓

Load Products

↓

Load Orders

↓

Load Admin

↓

Load Payment
```

Large initial load.

---

With Lazy Loading

```text
Application Starts

↓

Only Login Loads

↓

User opens Products

↓

Products Module Loads

↓

User opens Admin

↓

Admin Module Loads
```

Advantages

* Faster startup
* Better performance
* Reduced initial bundle size

---

# Standalone Components (Modern Angular)

Angular now supports **standalone components**, reducing the need for NgModules.

Example

```typescript
@Component({
  selector: 'app-home',
  standalone: true,
  imports: [],
  templateUrl: './home.component.html'
})
export class HomeComponent {}
```

Instead of declaring the component inside an `AppModule`, the component declares its own dependencies.

**Interview Tip:** Know both approaches. Many existing projects use NgModules, while new projects increasingly adopt standalone components.

---

# Real Project Example

Imagine an Employee Management System.

```text
AppModule

│
├── LoginModule
│      ├── LoginComponent
│      ├── RegisterComponent
│
├── EmployeeModule
│      ├── EmployeeListComponent
│      ├── EmployeeDetailsComponent
│
├── AdminModule
│      ├── DashboardComponent
│      ├── ReportsComponent
│
├── SharedModule
│      ├── HeaderComponent
│      ├── FooterComponent
│      ├── LoaderComponent
│
└── CoreModule
       ├── AuthService
       ├── LoggerService
```

This modular organization makes the application easier to maintain and extend.

---

# Best Practices

✅ Keep each module focused on a single feature.

✅ Place reusable UI elements in a `SharedModule`.

✅ Keep singleton services in a `CoreModule` or use `providedIn: 'root'`.

✅ Use lazy loading for large features.

✅ Avoid importing `BrowserModule` into feature modules.

✅ Prefer standalone components for new Angular applications when appropriate.

---

# Frequently Asked Interview Questions

### 1. What is a module in Angular?

A module is a class decorated with `@NgModule` that organizes related components, directives, pipes, and services into a cohesive unit.

---

### 2. What is AppModule?

`AppModule` is the root module that bootstraps the Angular application and loads the root component.

---

### 3. What is the purpose of `declarations`?

It registers the components, directives, and pipes that belong to the module.

---

### 4. What is the purpose of `imports`?

It brings functionality from other modules into the current module.

---

### 5. What is the purpose of `providers`?

It registers services for Angular's Dependency Injection system. Many services today instead use `providedIn: 'root'`.

---

### 6. What is the purpose of `bootstrap`?

It identifies the root component Angular should create first, usually `AppComponent`.

---

### 7. What is the difference between BrowserModule and CommonModule?

* `BrowserModule` is imported only by the root module.
* `CommonModule` is imported by feature modules and provides common directives like `*ngIf` and `*ngFor`.

---

### 8. What is lazy loading?

Lazy loading delays loading feature modules until the user navigates to them, improving startup performance.

---

### 9. What is a Shared Module?

A module containing reusable components, directives, and pipes that are shared across multiple feature modules.

---

### 10. What are standalone components?

Standalone components are components that manage their own dependencies and don't require an NgModule declaration,
simplifying application structure in modern Angular.

---

# Quick Revision Sheet

| Topic                 | Key Point                            |
| --------------------- | ------------------------------------ |
| Module                | Groups related Angular functionality |
| `@NgModule`           | Decorator that configures a module   |
| `declarations`        | Components, directives, pipes        |
| `imports`             | Other modules used by this module    |
| `providers`           | Services for Dependency Injection    |
| `bootstrap`           | Root component to launch             |
| BrowserModule         | Root module only                     |
| CommonModule          | Feature modules                      |
| SharedModule          | Reusable UI components               |
| CoreModule            | Singleton services                   |
| Lazy Loading          | Loads features on demand             |
| Standalone Components | Modern alternative to NgModules      |

---

# Practice Exercises

1. Create an `EmployeeModule` containing:

   * `EmployeeListComponent`
   * `EmployeeDetailsComponent`

2. Create a `SharedModule` containing:

   * `HeaderComponent`
   * `FooterComponent`

3. Import `SharedModule` into `EmployeeModule` and use the `HeaderComponent`.

4. Explain why `CommonModule` is used in feature modules instead of `BrowserModule`.

5. Describe how lazy loading would improve the performance of an e-commerce application.

Mastering these concepts gives you a solid understanding of how Angular applications are organized, and prepares you for
many common Angular interview questions. The next phase—**Templates & Data Binding**—builds directly on this by showing how components display data and respond to user interactions.
````

````python
# Angular Services & Dependency Injection (DI) – Complete Interview Guide

## 1. What is a Service?

### Answer

A **Service** is a TypeScript class that contains reusable business logic, data access logic, or shared functionality that can be used by multiple components.

Instead of writing the same logic in different components, we place it in a service and reuse it.

### Example

```typescript
@Injectable({
  providedIn: 'root'
})
export class EmployeeService {

  employees = ["John", "David", "Alice"];

  getEmployees() {
    return this.employees;
  }

}
```

### Interview Tip

Components should focus on the UI, while Services should contain business logic.

---

## 2. Why do we use Services?

### Answer

Services help keep the application organized by separating business logic from UI logic.

### Advantages

* Code Reusability
* Separation of Concerns
* Easy Maintenance
* Shared Data
* API Calls
* Better Testing
* Loose Coupling

### Real-World Example

Suppose both **EmployeeComponent** and **DashboardComponent** need employee data.

Instead of writing the same API call in both components, create an `EmployeeService` and inject it wherever needed.

---

## 3. Can two components use the same Service?

### Answer

Yes.

Multiple components can inject and use the same service.

If the service is provided using `providedIn: 'root'`, Angular creates only one shared instance (Singleton).

### Example

```text
EmployeeComponent
        │
        │
        ▼
EmployeeService
        ▲
        │
DashboardComponent
```

Both components use the same service instance.

---

## 4. How do Services communicate between Components?

### Answer

Services act as a shared data source between components.

A common approach is to use **RxJS Subjects** or **BehaviorSubjects**.

### Example

```typescript
@Injectable({
  providedIn: 'root'
})
export class MessageService {

  private message = new BehaviorSubject<string>("");

  currentMessage = this.message.asObservable();

  updateMessage(msg: string) {
    this.message.next(msg);
  }

}
```

Component A updates the message.

Component B subscribes and automatically receives updates.

### Interview Tip

For sibling component communication, using a shared service with `BehaviorSubject` is a common Angular pattern.

---

## 5. Difference between Component and Service

| Component                     | Service                                           |
| ----------------------------- | ------------------------------------------------- |
| Controls the UI               | Contains business logic                           |
| Has HTML, CSS, and TypeScript | Only TypeScript                                   |
| Handles user interaction      | Handles data, API calls, validation, calculations |
| Declared using `@Component`   | Declared using `@Injectable`                      |

### Interview Answer

A Component manages the user interface, while a Service manages reusable business logic and shared functionality.

---

# Dependency Injection (DI)

---

## 6. What is Dependency Injection?

### Answer

Dependency Injection (DI) is a design pattern where Angular automatically creates and provides required objects (dependencies) instead of the class creating them manually.

### Without DI

```typescript
EmployeeService service = new EmployeeService();
```

### With DI

```typescript
constructor(private service: EmployeeService) {}
```

Angular creates the object automatically.

### Advantages

* Loose Coupling
* Better Testing
* Code Reusability
* Easier Maintenance

---

## 7. Explain Dependency Injection with an Example

### Service

```typescript
@Injectable({
  providedIn: 'root'
})
export class EmployeeService {

  getMessage() {
    return "Welcome";
  }

}
```

### Component

```typescript
export class EmployeeComponent {

  constructor(private service: EmployeeService) {}

}
```

Angular automatically injects `EmployeeService` into the component.

### Interview Answer

Instead of creating objects manually using `new`, Angular's Dependency Injection container creates and supplies the required service.

---

## 8. What is `@Injectable()`?

### Answer

`@Injectable()` is a decorator that marks a class as available for Dependency Injection.

Example

```typescript
@Injectable({
  providedIn: 'root'
})
```

It tells Angular that the class can participate in the DI system.

---

## 9. What is Constructor Injection?

### Answer

Constructor Injection is the process of receiving dependencies through the class constructor.

Example

```typescript
constructor(private employeeService: EmployeeService) {}
```

Angular automatically injects the required service.

### Interview Tip

Constructor Injection is the most common form of Dependency Injection in Angular.

---

## 10. Why should we avoid using `new EmployeeService()`?

### Answer

Creating services manually bypasses Angular's Dependency Injection system.

### Problems

* Multiple service instances
* No Singleton behavior
* Harder to test
* Tightly coupled code
* Angular cannot manage the service lifecycle

### Correct Approach

```typescript
constructor(private service: EmployeeService) {}
```

### Wrong Approach

```typescript
const service = new EmployeeService();
```

---

## 11. What is a Singleton Service?

### Answer

A Singleton Service means Angular creates only one instance of the service for the entire application.

All components share the same instance.

### Benefits

* Shared state
* Lower memory usage
* Consistent data
* Better performance

---

## 12. What does `providedIn: 'root'` mean?

### Answer

```typescript
@Injectable({
  providedIn: 'root'
})
```

This tells Angular to register the service with the root injector.

As a result:

* Only one instance is created.
* The service is available throughout the application.
* You don't need to add it to the `providers` array of a module.

### Interview Answer

`providedIn: 'root'` creates a singleton service that is available application-wide through Angular's root injector.

---

# Quick Revision Table

| Question                            | Short Answer                                                         |
| ----------------------------------- | -------------------------------------------------------------------- |
| What is a Service?                  | Reusable business logic class                                        |
| Why use Services?                   | Reusability, maintainability, API calls, shared logic                |
| Can two components use one Service? | Yes, especially when it's a singleton                                |
| Component vs Service                | UI vs Business Logic                                                 |
| What is Dependency Injection?       | Angular automatically provides required objects                      |
| What is `@Injectable()`?            | Marks a class as injectable                                          |
| Constructor Injection               | Inject dependencies through the constructor                          |
| Why avoid `new`?                    | Bypasses Angular DI and creates unnecessary instances                |
| Singleton Service                   | One shared service instance across the application                   |
| `providedIn: 'root'`                | Registers the service with the root injector and creates a singleton |
````

````python
Here are the answers to all the Angular Directives and Routing questions.

## Directives

**1. What is a Directive?**
A directive is a class in Angular that adds custom behavior to elements in the DOM. It's marked with the `@Directive` decorator (components are actually a special kind of directive with a template attached via `@Component`).

**2. What are the different types of Directives?**
- **Component Directives** — directives with a template (`@Component`); the most common type.
- **Structural Directives** — change the DOM layout by adding/removing elements (e.g., `*ngIf`, `*ngFor`, `*ngSwitch`).
- **Attribute Directives** — change the appearance or behavior of an existing element without altering the DOM structure (e.g., `ngClass`, `ngStyle`, or custom ones like a highlight directive).

**3. Explain `*ngIf`**
A structural directive that conditionally adds or removes an element from the DOM based on a boolean expression.
```html
<div *ngIf="isLoggedIn">Welcome back!</div>
```
If `isLoggedIn` is `false`, the element is completely removed from the DOM (not just hidden) — Angular also destroys/recreates the component instance and its bindings each time the condition flips.

**4. Difference between `*ngIf` and `hidden`**
- `*ngIf` **adds/removes** the element from the DOM entirely. If false, no DOM node exists, no bindings/change detection run on it, and any component inside is destroyed (losing its state).
- `[hidden]` just toggles the CSS `display: none` property — the element stays in the DOM, still consumes memory, and its component state is preserved even when hidden.
Use `*ngIf` when the content is expensive to render or should be fully removed (e.g., large lists, forms with side effects); use `[hidden]` for cheap, frequent toggling where you want to preserve state.

**5. Explain `*ngFor`**
A structural directive used to loop over a collection and render a template once per item.
```html
<li *ngFor="let item of items; let i = index; trackBy: trackByFn">
  {{ i }} - {{ item.name }}
</li>
```
Local variables available: `index`, `first`, `last`, `even`, `odd`. `trackBy` is important for performance — it tells Angular how to identify each item so it can avoid destroying/recreating DOM nodes unnecessarily when the list updates.

**6. Explain `*ngSwitch`**
A structural directive group used for conditional rendering among multiple options, similar to a `switch` statement.
```html
<div [ngSwitch]="userRole">
  <p *ngSwitchCase="'admin'">Admin Panel</p>
  <p *ngSwitchCase="'user'">User Dashboard</p>
  <p *ngSwitchDefault>Guest View</p>
</div>
```
`[ngSwitch]` is bound to the container element, while `*ngSwitchCase` and `*ngSwitchDefault` are applied to the child elements to be conditionally shown.

**7. Explain `ngClass`**
An attribute directive used to dynamically add/remove CSS classes based on expressions.
```html
<div [ngClass]="{'active': isActive, 'disabled': isDisabled}">Content</div>
<!-- or -->
<div [ngClass]="isActive ? 'active' : 'inactive'"></div>
<!-- or with an array -->
<div [ngClass]="['class1', 'class2']"></div>
```

**8. Explain `ngStyle`**
An attribute directive used to dynamically set inline CSS styles based on expressions.
```html
<div [ngStyle]="{'color': isError ? 'red' : 'green', 'font-size.px': fontSize}">
  Status message
</div>
```

**9. Difference between `ngClass` and `ngStyle`**
- `ngClass` toggles **predefined CSS classes** — better for maintainability since styling logic stays in your stylesheet.
- `ngStyle` sets **inline styles directly** — useful for truly dynamic values computed at runtime (like a progress-bar width based on a variable), where defining a class for every possible value isn't practical.
Generally, `ngClass` is preferred when possible since it keeps styling separated from logic.

**10. Can we create Custom Directives?**
Yes. Use the Angular CLI (`ng generate directive myDirective`) or write one manually with `@Directive`. Example — a simple highlight directive:
```typescript
@Directive({
  selector: '[appHighlight]'
})
export class HighlightDirective {
  constructor(private el: ElementRef) {}

  @HostListener('mouseenter') onMouseEnter() {
    this.el.nativeElement.style.backgroundColor = 'yellow';
  }

  @HostListener('mouseleave') onMouseLeave() {
    this.el.nativeElement.style.backgroundColor = null;
  }
}
```
```html
<p appHighlight>Hover over me!</p>
```

**11. Why is `*` used before `*ngIf` and `*ngFor`?**
The `*` is syntactic sugar that tells Angular "this is a structural directive." Behind the scenes, Angular desugars it into an `<ng-template>` wrapping the element, since structural directives work by adding/removing whole template chunks from the DOM rather than just modifying an existing element.
```html
<div *ngIf="show">Hello</div>
<!-- desugars to -->
<ng-template [ngIf]="show">
  <div>Hello</div>
</ng-template>
```
Attribute directives (`ngClass`, `ngStyle`) don't need the `*` because they modify an existing element in place rather than adding/removing it from the DOM.

## Routing

**What is Routing?**
Angular's mechanism for navigating between different views/components within a single-page application (SPA) without a full page reload, based on changes to the URL.

**Why Routing?**
It enables SPA behavior — users can navigate between "pages" (really just different component views) using bookmarkable, shareable URLs, browser back/forward buttons work naturally, and only the necessary parts of the page re-render instead of reloading the entire app from the server. It also enables lazy loading of feature modules for better performance.

**What is `RouterModule`?**
The Angular module that provides the routing framework — including directives like `router-outlet` and `routerLink`, and services like `Router` and `ActivatedRoute`. You configure it with a set of routes:
```typescript
const routes: Routes = [
  { path: 'home', component: HomeComponent },
  { path: 'about', component: AboutComponent }
];

@NgModule({
  imports: [RouterModule.forRoot(routes)],
  exports: [RouterModule]
})
export class AppRoutingModule {}
```

**What is `router-outlet`?**
A placeholder directive that marks where the router should render the component matching the current route.
```html
<router-outlet></router-outlet>
```

**What is `routerLink`?**
A directive used to navigate to a route declaratively in a template, instead of setting a plain `href` (which would cause a full page reload).
```html
<a routerLink="/home">Home</a>
<a [routerLink]="['/user', userId]">View User</a>
```

**Difference between `href` and `routerLink`**
- `href` triggers a **full page reload**, re-downloading the entire app and losing SPA state — you don't want this for in-app navigation.
- `routerLink` uses Angular's router to swap views **without reloading the page**, preserving app state and being much faster, since the router just updates the `router-outlet` content based on the URL change.

**What is `ActivatedRoute`?**
An injectable service that gives you access to information about the route currently associated with a loaded component — including route parameters, query parameters, data, and the route's static configuration.
```typescript
constructor(private route: ActivatedRoute) {
  this.route.params.subscribe(params => {
    this.id = params['id'];
  });
}
```

**Route Parameters**
Dynamic segments of a route's path, used to pass data as part of the URL itself (e.g., an ID).
```typescript
{ path: 'user/:id', component: UserComponent }
```
```typescript
this.route.paramMap.subscribe(params => {
  this.userId = params.get('id');
});
```
URL example: `/user/42`

**Query Parameters**
Optional key-value pairs appended to a URL after a `?`, typically used for things like filters, sorting, or pagination — not part of the route's path structure.
```typescript
this.router.navigate(['/products'], { queryParams: { category: 'shoes', sort: 'price' } });
```
URL: `/products?category=shoes&sort=price`
```typescript
this.route.queryParams.subscribe(params => {
  this.category = params['category'];
});
```

**What is Lazy Loading?**
A technique where feature modules are loaded **on demand** — only when the user navigates to a route that needs them — instead of loading the entire application's code upfront at startup.
```typescript
{ path: 'admin', loadChildren: () => import('./admin/admin.module').then(m => m.AdminModule) }
```

**Advantages of Lazy Loading**
- Reduces the **initial bundle size**, so the app loads and becomes interactive faster.
- Improves performance for large applications by splitting code into smaller chunks loaded only when needed.
- Better resource utilization — users who never visit certain features never download that code at all.

**What are Route Guards?**
Interfaces that let you control whether navigation to (or away from) a route is allowed to proceed — used for things like authentication checks, unsaved-changes warnings, or permission checks, executed before the router activates or deactivates a route.

**Types of Route Guards**
- **`CanActivate`** — controls whether a route can be entered (e.g., blocking unauthenticated users).
- **`CanActivateChild`** — same as above, but for child routes.
- **`CanDeactivate`** — controls whether the user can leave a route (e.g., warning about unsaved form changes).
- **`Resolve`** — pre-fetches data before the route is activated, so the component has data ready immediately on load.
- **`CanLoad`** (older API) / **`CanMatch`** (newer, Angular 14.1+) — controls whether a lazily-loaded module is allowed to load at all, useful for preventing unauthorized users from even downloading a feature module's code.
````

````python
# Angular Observables & RxJS – Complete Interview Guide

## 1. What is an Observable?

### Answer

An **Observable** is an RxJS object that represents a stream of asynchronous data over time. It can emit **zero, one, or multiple values**, making it ideal for handling events, HTTP requests, and real-time updates.

### Example

```typescript
this.http.get<Employee[]>('/api/employees')
    .subscribe(data => console.log(data));
```

### Uses

* HTTP Requests
* User Events
* WebSockets
* Timers
* Real-time Notifications

---

## 2. Difference between Observable and Promise

| Observable                     | Promise                         |
| ------------------------------ | ------------------------------- |
| Emits multiple values          | Emits only one value            |
| Lazy (starts on subscribe)     | Eager (starts immediately)      |
| Can be cancelled (unsubscribe) | Cannot be cancelled             |
| Supports RxJS operators        | Limited chaining with `.then()` |
| Used extensively in Angular    | Native JavaScript feature       |

### Interview Tip

Angular uses **Observables** because they are more powerful and flexible than Promises.

---

## 3. Why does HttpClient return Observable?

### Answer

Angular's `HttpClient` returns an Observable because it supports:

* Asynchronous operations
* Cancellation
* Error handling
* RxJS operators (`map`, `filter`, `switchMap`, etc.)
* Better composition of asynchronous tasks

---

## 4. What is `subscribe()`?

### Answer

`subscribe()` starts the Observable and allows you to receive the emitted values.

### Example

```typescript
this.http.get('/api/users').subscribe(data => {
    console.log(data);
});
```

Without `subscribe()`, the Observable does not execute.

---

## 5. Explain `next()`

### Answer

`next()` is called whenever the Observable emits a value.

### Example

```typescript
observable.subscribe({
    next: data => console.log(data)
});
```

---

## 6. Explain `error()`

### Answer

`error()` is executed if the Observable encounters an error.

### Example

```typescript
observable.subscribe({
    error: err => console.log(err)
});
```

Used for displaying error messages or logging failures.

---

## 7. Explain `complete()`

### Answer

`complete()` is called when the Observable finishes emitting values successfully.

### Example

```typescript
observable.subscribe({
    complete: () => console.log("Completed")
});
```

---

## 8. What is `pipe()`?

### Answer

`pipe()` is used to chain RxJS operators together to transform or process Observable data.

### Example

```typescript
this.http.get('/api/users')
    .pipe(
        map(data => data),
        catchError(err => throwError(() => err))
    );
```

---

## 9. What is `switchMap()`?

### Answer

`switchMap()` switches to a new Observable and automatically cancels the previous one.

### Common Uses

* Search suggestions
* Auto-complete
* API calls based on user input

### Interview Tip

`switchMap()` prevents unnecessary API calls by cancelling outdated requests.

---

## 10. What is `catchError()`?

### Answer

`catchError()` is an RxJS operator used to catch and handle Observable errors gracefully.

### Example

```typescript
.pipe(
    catchError(error => {
        console.error(error);
        return throwError(() => error);
    })
)
```

---

## 11. Why unsubscribe?

### Answer

Unsubscribing stops the Observable and prevents unnecessary background processing.

### Benefits

* Prevents memory leaks
* Improves performance
* Stops unwanted updates

---

## 12. What is a Memory Leak?

### Answer

A memory leak occurs when objects remain in memory even though they are no longer needed.

In Angular, forgetting to unsubscribe from long-lived Observables can keep components alive in memory after they are destroyed.

### Solution

Always unsubscribe in `ngOnDestroy()` or use operators like `takeUntil()` or Angular utilities such as `AsyncPipe` where appropriate.

---

# Quick Revision

| Question              | Short Answer                                              |
| --------------------- | --------------------------------------------------------- |
| Observable            | Stream of asynchronous data                               |
| Promise vs Observable | Single value vs Multiple values                           |
| HttpClient            | Returns Observable                                        |
| subscribe()           | Starts the Observable                                     |
| next()                | Receives emitted values                                   |
| error()               | Handles errors                                            |
| complete()            | Runs when Observable finishes                             |
| pipe()                | Chains RxJS operators                                     |
| switchMap()           | Switches to a new Observable and cancels the previous one |
| catchError()          | Handles Observable errors                                 |
| unsubscribe()         | Stops Observables                                         |
| Memory Leak           | Unreleased resources due to active subscriptions          |
````

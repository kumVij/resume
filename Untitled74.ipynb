````python
# Git Interview Prep — Commands Explained

---

## 1. Setup & Configuration

```bash
git init                      # Turn the current folder into a new Git repo (creates .git/)
git clone <url>                # Copy a remote repo (and its full history) to your machine
git config --global user.name "Name"
git config --global user.email "email@example.com"
                                # Set identity used in your commit metadata
git config --list              # Show current config settings
```

---

## 2. The Three Areas (core mental model)

Git tracks files across three states — almost every command below is really about moving changes between these:
1. **Working directory** — your actual files on disk.
2. **Staging area / index** — a draft of what will go into the next commit.
3. **Repository (.git)** — the committed history.

```
working directory --(git add)--> staging area --(git commit)--> repository
```

---

## 3. Basic Snapshot Commands

```bash
git status                     # Show what's changed, staged, or untracked
git add <file>                 # Stage a specific file's changes
git add .                      # Stage everything in the current directory
git commit -m "message"        # Save staged changes as a new commit in history
git commit -am "message"       # Stage + commit all TRACKED file changes in one step (skips new/untracked files)
git log                        # Show commit history (newest first)
git log --oneline --graph      # Compact, visual view of history and branches
git diff                       # Show unstaged changes (working dir vs staging)
git diff --staged              # Show staged changes (staging vs last commit)
git show <commit>               # Show the changes introduced by a specific commit
```

---

## 4. Branching

```bash
git branch                     # List local branches (current one marked with *)
git branch <name>              # Create a new branch (doesn't switch to it)
git checkout <name>             # Switch to an existing branch
git checkout -b <name>          # Create AND switch to a new branch in one step
git switch <name>                # Modern alternative to checkout for switching branches
git switch -c <name>              # Modern alternative to checkout -b
git branch -d <name>             # Delete a branch (safe — refuses if unmerged commits exist)
git branch -D <name>             # Force-delete a branch, even with unmerged commits
```
**Interview point:** branches are just lightweight, movable pointers to a commit — not full copies of the codebase, which is why Git branching is cheap and fast compared to some other VCS.

---

## 5. Merging & Rebasing

```bash
git merge <branch>              # Combine the named branch's changes into your current branch
git rebase <branch>              # Replay your current branch's commits on top of the named branch
git rebase -i <commit>            # Interactive rebase — reorder, squash, edit, or drop commits
```

**Merge vs Rebase (classic interview question):**
| Merge | Rebase |
|---|---|
| Creates a new "merge commit" tying two histories together | Rewrites commit history — moves your commits to start from the tip of the target branch |
| Preserves exact history, including all branch points | Produces a linear, cleaner history |
| Safe on shared/public branches | **Never rebase commits that have already been pushed/shared** — it rewrites history and breaks collaborators' copies |

```bash
git merge --no-ff <branch>        # Force a merge commit even if a fast-forward were possible (keeps branch history visible)
```

---

## 6. Undoing Changes

```bash
git restore <file>                # Discard uncommitted changes in a file (working dir → last commit state)
git restore --staged <file>       # Unstage a file (keeps the changes, just removes from staging)
git checkout -- <file>            # Older equivalent of `git restore <file>`
git reset <file>                   # Older equivalent of `git restore --staged <file>`
git reset --soft <commit>          # Move HEAD to <commit>, keep changes staged
git reset --mixed <commit>         # Move HEAD to <commit>, keep changes in working dir but unstaged (default mode)
git reset --hard <commit>          # Move HEAD to <commit>, DISCARD all changes after it — destructive, use carefully
git revert <commit>                 # Create a NEW commit that undoes a previous commit — safe for shared/pushed history
```
**Interview point (reset vs revert):** `reset` rewrites history by moving the branch pointer backward — dangerous once pushed. `revert` adds a new commit that undoes changes, preserving history — the safe choice for shared branches.

---

## 7. Stashing

```bash
git stash                          # Temporarily shelve uncommitted changes, restore a clean working dir
git stash list                     # Show all stashed change sets
git stash pop                       # Reapply the most recent stash AND remove it from the stash list
git stash apply                      # Reapply the most recent stash but KEEP it in the stash list
git stash drop                        # Delete a stash without applying it
```
Use case: you're mid-work on a feature and need to quickly switch branches (e.g., for a hotfix) without committing half-done work.

---

## 8. Remote Repositories

```bash
git remote -v                        # List remote repos and their URLs (fetch/push)
git remote add origin <url>          # Link a local repo to a remote named "origin"
git fetch                             # Download remote commits/branches WITHOUT merging into your working branch
git pull                               # Equivalent to `git fetch` + `git merge` (or rebase, if configured) in one step
git pull --rebase                      # Fetch, then rebase your local commits on top instead of merging
git push                                # Upload local commits to the remote branch
git push -u origin <branch>             # Push and set the upstream tracking branch (so future `git push` needs no args)
git push --force                        # Overwrite remote history with your local branch — dangerous, avoid on shared branches
git push --force-with-lease             # Safer force-push: fails if the remote has commits you haven't seen yet
```
**Interview point (fetch vs pull):** `fetch` is non-destructive — it just updates your local view of the remote; `pull` actually integrates those changes into your current branch, which is why `pull` can cause conflicts and `fetch` never does.

---

## 9. Inspecting & Comparing

```bash
git log --author="name"              # Filter commits by author
git log -p                            # Show full diffs alongside each commit
git blame <file>                       # Show who last changed each line of a file, and in which commit
git diff <branch1> <branch2>           # Compare two branches
git tag <name>                         # Mark a specific commit (e.g., a release) with a readable label
```

---

## 10. Cherry-pick & Conflict Resolution

```bash
git cherry-pick <commit>                # Apply a single specific commit from another branch onto your current branch
```
Use case: you need just one bug fix from another branch, not the whole branch's history.

**Merge conflicts:** happen when Git can't automatically reconcile changes to the same lines. Git marks the conflicting sections in the file with `<<<<<<<`, `=======`, `>>>>>>>` markers; you manually edit to resolve, then:
```bash
git add <file>       # Mark the conflict as resolved
git commit             # Complete the merge (or `git rebase --continue` if mid-rebase)
```

---

## 11. Frequently Asked Conceptual Questions

1. **What's the difference between `git merge` and `git rebase`?** See section 5 table — merge preserves full history with a merge commit; rebase rewrites history into a linear sequence.
2. **What's the difference between `git fetch` and `git pull`?** `fetch` downloads without merging; `pull` downloads and merges (or rebases) automatically.
3. **What's the difference between `git reset` and `git revert`?** `reset` moves the branch pointer and can discard history (dangerous on shared branches); `revert` adds a new commit undoing changes (safe on shared branches).
4. **What is HEAD?** A pointer to the current commit/branch you're working on — "where you are" in the repo.
5. **What's a detached HEAD state?** When you check out a specific commit (not a branch) — new commits made here aren't attached to any branch and can be lost unless you create a branch from that point.
6. **What is `.gitignore`?** A file listing patterns of files/folders Git should never track (build artifacts, `node_modules`, secrets, `.env` files).
7. **What's the difference between a fast-forward merge and a true merge commit?** Fast-forward just moves the branch pointer forward when there's no divergent history; a true merge commit is created when both branches have unique commits, tying the histories together.
8. **How do you resolve a merge conflict?** Manually edit the conflict markers in affected files, `git add` the resolved files, then `git commit` (or `git rebase --continue`).
9. **What does `git squash` (via interactive rebase) do, and why use it?** Combines multiple commits into one, typically to clean up a messy feature branch history before merging to main.
10. **Why should you avoid `git push --force` on a shared branch?** It overwrites remote history, which can silently delete teammates' commits; `--force-with-lease` is the safer alternative since it fails if someone else has pushed since you last fetched.

---

## 12. Quick Self-Check
Before an interview, you should be able to, without notes:
- Explain the three Git areas (working dir / staging / repo) and which command moves changes between each.
- Walk through resolving a merge conflict from marker to commit.
- Explain reset (soft/mixed/hard) vs revert with a concrete scenario for each.
- Explain why you'd rebase a feature branch before opening a PR, but never rebase `main`.
- Explain stash and cherry-pick with a real use case for each.
````
